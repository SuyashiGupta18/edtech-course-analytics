```python
import pandas as pd

df = pd.read_csv('edtech_course_data.csv')

print(df.shape)
df.head()
```

```python
df['payment_method'] = df['payment_method'].fillna('Unknown')
```

```python
df['time_spent_hours'] = df['time_spent_hours'].fillna(df['time_spent_hours'].median())
```

```python
df.isnull().sum()
```

```python
df['quiz_avg_score'].unique()[:20]
```

```python
# Step 1: Create a separate, clear flag column BEFORE touching the numeric column
df['quiz_attempted'] = df['num_quizzes_taken'] > 0

# Step 2: Replace the text 'Not Attempted' back with proper NaN, restoring the column to pure numbers
df['quiz_avg_score'] = df['quiz_avg_score'].replace('Not Attempted', pd.NA)
df['quiz_avg_score'] = pd.to_numeric(df['quiz_avg_score'])

# Step 3: Verify
print(df['quiz_avg_score'].dtype)
print(df['quiz_avg_score'].isnull().sum())
df['quiz_attempted'].value_counts()
```

```python
print("Duplicates found:", df.duplicated().sum())
df = df.drop_duplicates()
print("New shape:", df.shape)
```

```python
print(df['category'].unique())

df['category'] = df['category'].str.strip().str.lower()

category_map = {
    'digital marketing': 'Digital Marketing',
    'personal development': 'Personal Development',
    'business & finance': 'Business & Finance',
    'business and finance': 'Business & Finance',
    'web dev': 'Web Development',
    'web development': 'Web Development',
    'it and software': 'IT & Software',
    'it & software': 'IT & Software',
    'cloud computing': 'Cloud Computing',
    'data  science': 'Data Science',
    'data science': 'Data Science',
    'datascience': 'Data Science',
    'graphic design': 'Graphic Design',
}

df['category'] = df['category'].map(category_map)
print(df['category'].unique())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='category', y='completion_pct', estimator='mean', errorbar=None)
plt.title('Average Completion % by Course Category')
plt.xlabel('Category')
plt.ylabel('Average Completion %')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
dropout_by_cat = df.groupby('category')['dropout_flag'].apply(lambda x: (x == 'Yes').mean() * 100).sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=dropout_by_cat.index, y=dropout_by_cat.values, color='indianred')
plt.title('Dropout Rate (%) by Course Category')
plt.xlabel('Category')
plt.ylabel('Dropout Rate (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(data=df, x='completion_status', y='time_spent_hours', estimator='mean', errorbar=None,
            order=['Not Started', 'In Progress', 'Completed'])
plt.title('Average Time Spent by Completion Status')
plt.xlabel('Completion Status')
plt.ylabel('Average Time Spent (hours)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data=df, x='completion_pct', bins=20, kde=True)
plt.title('Distribution of Completion Percentage')
plt.xlabel('Completion %')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='dropout_flag', y='quiz_avg_score')
plt.title('Quiz Score Distribution: Dropout vs. Not')
plt.xlabel('Dropped Out?')
plt.ylabel('Quiz Average Score')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
numeric_cols = ['age', 'price_inr', 'time_spent_hours', 'completion_pct', 'num_quizzes_taken', 'quiz_avg_score', 'rating']
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))
df['completion_status'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['#4CAF50', '#FFC107', '#F44336'])
plt.title('Overall Completion Status Breakdown')
plt.ylabel('')
plt.tight_layout()
plt.show()
```
